# Example Movement graph for Paul Skenes
This graph I made to give visotors of my website, and those who are unfamiliar with a baseball an introductory into how pitch movement is measured and classified so that different pitches can be distinguished.

```python
# Assign each pitches statistics
ps_fastball = {
    'velocity': 97.3,
    'usage': .36,
    'HB': 14.6,
    'VB': 10.7
}
ps_sinker = {
    'velocity': 94.0,
    'usage': .15,
    'HB': 18,
    'VB': 6
}
ps_curveball = {
    'velocity': 81.8,
    'usage': .02,
    'HB': -10,
    'VB': -10
}
ps_sweeper = {
    'velocity': 83.8,
    'usage': .16,
    'HB': -13,
    'VB': 6
}
ps_splitter = {
    'velocity': 93.6,
    'usage': .14,
    'HB': 12.9,
    'VB': .6
}
ps_changeup = {
    'velocity': 89.0,
    'usage': .18,
    'HB': 17,
    'VB': 6.8
}
```

```python
# Combine pitch statistics into a DataFrame
pitch_data = [
    {'pitch_name': 'Fastball', **ps_fastball},
    {'pitch_name': 'Sinker', **ps_sinker},
    {'pitch_name': 'Curveball', **ps_curveball},
    {'pitch_name': 'Sweeper', **ps_sweeper},
    {'pitch_name': 'Splitter', **ps_splitter},
    {'pitch_name': 'Changeup', **ps_changeup}
]
pitch_df = pd.DataFrame(pitch_data)

# Plot and create the scatter plot
fig = px.scatter(
    pitch_df,
    x='HB',
    y='VB',
    hover_data=['velocity', 'usage'],
    title='Paul Skenes Pitch Movement Profile',
    labels={
        'HB': 'Horizontal Break (inches)',
        'VB': 'Vertical Break (inches)'
    },
    height=600,
    width=800
)

fig.update_traces(marker=dict(size=15))

for index, row in pitch_df.iterrows():
    yshift_val = 10
    if row['pitch_name'] == 'Sinker':
        yshift_val = -15
    elif row['pitch_name'] == 'Changeup':
        yshift_val = 15

    fig.add_annotation(
        x=row['HB'],
        y=row['VB'],
        text=row['pitch_name'],
        showarrow=False,
        yshift=yshift_val,
        font=dict(color='black', size=12, weight='bold')
    )

fig.add_shape(
    type="line",
    x0=-25, x1=25, y0=0, y1=0,
    line=dict(color="grey", width=1, dash="dash")
)
fig.add_shape(
    type="line",
    x0=0, x1=0, y0=-25, y1=25,
    line=dict(color="grey", width=1, dash="dash")
)

# Layout
fig.update_layout(
    hovermode="closest",
    xaxis_title="Horizontal Break (inches)",
    yaxis_title="Vertical Break (inches)",
    xaxis_range=[-25, 25],
    yaxis_range=[-25, 25]
)

fig.show()

fig.write_html("paul_skenes.html")
```

# One New Visualization Technique: Informational strikezone

```python
import plotly.graph_objects as go

groups = [
    {
        "group":           "Four Seam and Changeup/Splitter",
        "x":               0.25,
        "y":               0.80,
        "notes":           "Pitchers in this group utilize their 4 seam and offspeed tunnel to get hitters out.<br>Splitters and Changeups are particularly effective against opposite handed batters allowing for these pitchers to be a matchup nightmare.",
        "example_pitcher": "Yoshinobu Yamamoto",
    },
    {
        "group":           "Changeup Specialist",
        "x":               0.60,
        "y":               0.65,
        "notes":           "Pitchers in this group throw their changeups/splitters at an astonishing 20% rate or higher, way above league average.<br>These pitchers rely on outlier offspeed shapes to carve their way through a lineup.",
        "example_pitcher": "Logan Webb",
    },
    {
        "group":           "Big Arsenal",
        "x":               0.45,
        "y":               0.45,
        "notes":           "This grouping of pitchers don't rely on any one pitch in particular to get opposing hitters out.<br>All pitchers in this group utilize atleast 5+ pitches they throw at atleast a 10% usage rate each.",
        "example_pitcher": "Bryan Woo",
    },
    {
        "group":           "Cutter/Sinker",
        "x":               0.75,
        "y":               0.30,
        "notes":           "Cutters and sinkers are similar to 4 seam fastballs in velocity but move slightly toward and away from the batter.<br>This grouping of pitchers induces a lot of weak contact because of the firm and shorter shape of the pitches,<br> but struggles to produce big strikeout numbers.",
        "example_pitcher": "Garret Crochett",
    },
    {
        "group":           "4 Seam + 12/6",
        "x":               0.20,
        "y":               0.20,
        "notes":           "These pitchers are most successful when they are tunneling their primary breaking ball off of their high ride fastball.<br>They do play a dangerous game however as they are highly susceptible to fly balls and harder contact.<br>Many of these pitchers boast big whiff rates.",
        "example_pitcher": "Hunter Brown",
    },
    {
        "group":           "Primary Sinker",
        "x":               0.55,
        "y":               0.85,
        "notes":           "Pitchers who throw high dosages of sinkers induce lots of ground balls.<br>Sinker reliant pitchers also often couple their fastball with hard sliders and sweepers to tunnel off of one another.",
        "example_pitcher": "Ranger Suarez",
    },
    {
        "group":           "Lefty Changeup Specialist",
        "x":               0.85,
        "y":               0.55,
        "notes":           "Similar to the Changeup Specialist, this group of pitchers throws a large amount of changeups from the left side.<br>As majority of hitters in Major League Baseball are right handed, <br> this is a particularly successful group in putting up big soft contact and strikeout numbers to highly right handed lineups.",
        "example_pitcher": "Tarik Skubal",
    },

]

# Colors for each group
dot_colors = [
    "#1D9E75",  # teal
    "#D85A30",  # coral
    "#378ADD",  # blue
    "#BA7517",  # amber
    "#7F77DD",  # purple
    "#D4537E",  # pink
    "#639922",  # green
]

# Building the strikezone
fig = go.Figure()

fig.add_shape(
    type="rect",
    x0=0, y0=0, x1=1, y1=1,
    line=dict(color="gray", width=2, dash="dash"),
    fillcolor="rgba(200,220,255,0.06)",
)

# strikezone grid lines
for v in [1/3, 2/3]:
    fig.add_shape(type="line", x0=v, y0=0, x1=v, y1=1,
                  line=dict(color="gray", width=0.5, dash="dot"))
    fig.add_shape(type="line", x0=0, y0=v, x1=1, y1=v,
                  line=dict(color="gray", width=0.5, dash="dot"))

# Home plate
fig.add_trace(go.Scatter(
    x=[0.2, 0.8, 0.8, 0.5, 0.2, 0.2],
    y=[-0.22, -0.22, -0.12, -0.06, -0.12, -0.22],
    mode="lines", fill="toself",
    fillcolor="white", line=dict(color="gray", width=1.5),
    hoverinfo="skip", showlegend=False,
))



# Tracing for each group to display information
for i, g in enumerate(groups):
    color = dot_colors[i % len(dot_colors)]
    fig.add_trace(go.Scatter(
        x=[g["x"]],
        y=[g["y"]],
        mode="markers",
        name=g["group"],
        marker=dict(color=color, size=18, line=dict(color="white", width=1.5)),
        customdata=[[g["group"], g["notes"], g["example_pitcher"]]],
        hovertemplate=(
            "<b>%{customdata[0]}</b><br>" # Group name in bold
            "──────────────────────<br>" # Separator line
            "%{customdata[1]}<br>" # Notes, now with newlines handled by <br>
            "<br>" # Add an extra break for visual separation
            "<i>Example pitcher: %{customdata[2]}</i>"
            "<extra></extra>"
        ),
    ))

# Layout
fig.update_layout(
    title=dict(text="Pitcher Cluster Groups — Strike Zone", font=dict(size=20), x=0.5),
    xaxis=dict(
        range=[-0.3, 1.55],
        showgrid=False, zeroline=False, showticklabels=False,
        fixedrange=True,
    ),
    yaxis=dict(
        range=[-0.35, 1.15],
        showgrid=False, zeroline=False, showticklabels=False,
        fixedrange=True, scaleanchor="x", scaleratio=1,
    ),
    plot_bgcolor="rgba(245,247,250,1)",
    paper_bgcolor="white",
    legend=dict(
        title="Group",
        orientation="v",
        yanchor="middle", y=0.5,
        xanchor="left",  x=1.42,
        font=dict(size=11),
    ),
    margin=dict(l=20, r=220, t=70, b=20),
    width=860,
    height=600,
    hoverlabel=dict(bgcolor="white", font_size=13, bordercolor="lightgray", font=dict(color="black")),
)


fig.show()
fig.write_html("strikezone.html")
```

This visualization shows each of the 7 groups I created and provides insight into what each group was created for, as well as an example pitcher.

# K-BB%

```python
# K-BB rate analysis for mean, median, standard deviation
kbb_analysis = df.groupby('cluster_name')['k_minus_bb_percent'].agg(['mean', 'median', 'std', 'count']).sort_values(by='mean', ascending=False)
```

# xWOBA

```python
# xWOBA analysis for mean, median, standard deviation
xwoba_analysis = df.groupby('cluster_name')['xwoba'].agg(['mean', 'median', 'std', 'count']).sort_values(by='mean', ascending=False)
```

# Average Exit Velocity

```python
# Average Exit Velocity analysis for mean, median, standard deviation
exit_velocity_analysis = df.groupby('cluster_name')['exit_velocity_avg'].agg(['mean', 'median', 'std', 'count']).sort_values(by='mean', ascending=False)
```

# Whiff%

```python
# Whiff rate analysis for mean, median, standard deviation
whiff_analysis = df.groupby('cluster_name')['whiff_percent'].agg(['mean', 'median', 'std', 'count']).sort_values(by='mean', ascending=False)
```

# 4D Scatterplot to display xWOBA, K-BB%, Whiff%, and AVG Exit-Velocity

```python
# Mean columns created above from each statistics analysis
df_kbb_mean = kbb_analysis[['mean']].rename(columns={'mean': 'k_minus_bb_percent_mean'})
df_xwoba_mean = xwoba_analysis[['mean']].rename(columns={'mean': 'xwoba_mean'})
df_exit_velo_mean = exit_velocity_analysis[['mean']].rename(columns={'mean': 'exit_velocity_avg_mean'})
df_whiff_mean = whiff_analysis[['mean']].rename(columns={'mean': 'whiff_percent_mean'})

# Merge them into a single DataFrame
grouped_means_df = df_kbb_mean.merge(df_xwoba_mean, left_index=True, right_index=True)
grouped_means_df = grouped_means_df.merge(df_exit_velo_mean, left_index=True, right_index=True)
grouped_means_df = grouped_means_df.merge(df_whiff_mean, left_index=True, right_index=True)

# Reset index
grouped_means_df = grouped_means_df.reset_index()

# 3 Dimensional Scatter Plot using the grouped means
fig = px.scatter_3d(
    grouped_means_df,
    x='xwoba_mean',
    y='k_minus_bb_percent_mean',
    z='whiff_percent_mean',
    color='exit_velocity_avg_mean',
    size=[10] * len(grouped_means_df),  # Use a fixed size for the 7 points
    hover_name='cluster_name',
    hover_data={
        'xwoba_mean': ':.3f',
        'k_minus_bb_percent_mean': ':.2f',
        'exit_velocity_avg_mean': ':.1f',
        'whiff_percent_mean': ':.1f'
    },
    title='Pitcher Cluster Group Averages (2025)',
    labels={
        'xwoba_mean': 'xWOBA',
        'k_minus_bb_percent_mean': 'K - BB Rate',
        'exit_velocity_avg_mean': 'Average Exit Velo',
        'whiff_percent_mean': 'Whiff Rate'
    }
)

fig.update_layout(scene_camera=dict(eye=dict(x=1.8, y=1.8, z=0.8)), height=800)
fig.show()

fig.write_html("4dscatterplot.html")
```

This visualation is a 4D scatterplot that shows all of the important performance statistics I was using to measure the pitch arsenal groups.
